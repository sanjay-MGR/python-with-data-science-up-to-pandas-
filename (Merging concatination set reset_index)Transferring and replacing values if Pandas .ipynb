# Group By 

```python
cars_data.head(2)
```

```python
cars_data.columns
```

```python
# find the min,max,sum,mean,median,std of mileage and power group by location and brand

new_data_cars = cars_data.groupby(['location','brand'])[['mileage','power']].agg(['mean','std','sum','median','min','max']).reset_index()
```

```python
new_data_cars
```

```python
# find the min,sum,max for mileage and mean,std,median power group by location and brand
new_data_cars = cars_data.groupby(['location','brand'])[['mileage','power']].agg({
    'mileage':['min','max','sum'],'power':['mean','std','median']
}).reset_index()
```
