```python
data = pd.read_csv(r'cars_data_updated - cars_data_updated (1).csv')
```

```python
data.head(2)
```

```python
df = data.copy()
```

```python
#null value handel 
df.isna().sum()
```

```python
df.shape
```

```python
df.head(2)
```

```python
df_1 = df.copy()
```

```python
df.head(2)
```

```python
df.drop(['@'],axis = 1,inplace=True) # drop the column name '@'
```

```python
df.head(2)
```

```python
df.isna().sum() # check null value 
```

```python
df['mileage'] = df['mileage'].fillna(df['mileage'].mean()) #fill the null value of mileage with its mean 
```

```python
#average engine
e_m = df.engine.mean()
```

```python
df.engine = df.engine.fillna(e_m)#fill the null value of engine with its mean 
```

```python
#fill by bfill methode on power 
df.power = df.power.fillna(method = 'bfill')
```

```python
#fill by bfill methode on seats
df['seats'] = df['seats'].fillna(method = 'ffill')
```

```python
df.drop(['new_price'],axis =1,inplace = True) # drop new_price columns
```

```python
df['price'] = df.price.fillna(method = 'ffill')
```

```python
df.isna().sum()
```

```python
df.head(5)
```

# count plot

```python
sns.countplot(x = 'brand',data = df)
plt.xticks(rotation = 75)
plt.title('count plot of brand')
plt.show()
```

# line plot 


```python
df_l = df[['location','mileage']]
df_l.head(2)
```

```python
#plot 
plt.figure(figsize = (10,5))
sns.lineplot(x = 'location',y = 'mileage',data = df_l,marker = 'D')
plt.title('line plot between location and mileage')
plt.xticks(rotation = 75)
plt.xlabel('Location')
plt.ylabel('Mileage')
plt.show()
```

# Bar plot

```python
df_b = df[['location','mileage']]
df_b
# plot 
plt.figure(figsize = (10,5))
sns.barplot(x = 'location',y = 'mileage',data = df_b)
plt.title('bar plot between location and mileage')
plt.xticks(rotation = 75)
plt.xlabel('Location')
plt.ylabel('Mileage')
plt.show()
```

# scatter plot

```python
#scatter plot 
df_s = df[['brand','mileage']]
df_s
# plot 
plt.figure(figsize = (10,5))
sns.scatterplot(x = 'brand',y = 'mileage',data = df_s)
plt.title('scatter plot between brand and mileage')
plt.xticks(rotation = 75)
plt.xlabel('brand')
plt.ylabel('Mileage')
plt.show()
```

# Box plot 

```python
#box plot 
df_bx = df[['brand','mileage']]
df_bx
# plot 
plt.figure(figsize = (10,5))
sns.boxplot(x = 'brand',y = 'mileage',data = df_bx)
plt.title('Box plot between brand and mileage')
plt.xticks(rotation = 75)
plt.xlabel('brand')
plt.ylabel('Mileage')
plt.show()
```

# Histogram 


```python
sns.histplot(x = 'brand',data = df)
plt.xticks(rotation = 90)
plt.title('histogram of brand ')
plt.show()
```

```python
#also
sns.histplot(x = 'brand',kde = True,data = df)
plt.xticks(rotation = 90)
plt.show()
```

# kernel density Estimation(kde)plot(we can use only numeric variable here)

```python
median_power = df['power'].median()
median_power
```

```python
#kernel density Estimation(kde)plot(we can use only numeric variable here)
sns.kdeplot(x = 'power',data = df)
plt.axvline(median_power,linestyle='--',color = 'red')
plt.axhline(0.013,linestyle='--',color = 'green')
plt.xticks(rotation = 90)
plt.show()
```

# Violineplot

```python
# violine plot 
sns.violinplot(x='trans',y ='mileage',data = df,hue = 'trans')
plt.title(' violine plot  tans vs mileage')
plt.show()
```
