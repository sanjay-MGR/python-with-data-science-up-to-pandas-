```python
df.head(5)
```

# Resample Date

```python

daily_sales = df.resample('D').agg({'Sales':'sum'})
daily_sales
```

```python
daily_profit = df.resample('D').agg({'Profit':'sum'}).reset_index()
daily_profit
```

```python

#End of the month 
Mothly_end = df.resample('M').agg({'Profit':'sum','Sales':'sum'})
Mothly_end.head(5)
```

```python
#start of the month 
Mothly_start = df.resample('MS').agg({'Profit':'sum','Sales':'sum'}).reset_index()
Mothly_start.head(5)
```

```python

#End of the quarter
q_end = df.resample('Q').agg({'Sales':['max','min']}).reset_index()
q_end.head(5)
```

```python
#start of the quarter
q_start = df.resample('QS').agg({'Sales':['max','min'],'Profit':['mean','sum']})
q_start.head(5)
```

```python

#end of the year 
year_end = df.resample('Y').agg({'Sales':['max','min']})
year_end.head(5)
```

```python
#start of the year 
year_start = df.resample('YS').agg({'Sales':['mean','sum'],'Profit':['max','min']})
year_start
```

```python
year_start.reset_index(inplace = True) # reset index
```

```python
year_start
```

```python
weekly_basis = df.resample('W').agg({'Sales':'sum','Profit':'sum'})
weekly_basis 
```

```python
data_monthly = df.resample('M').agg({'Sales':'sum','Profit':'sum'})
data_monthly.head(5)
```

```python
data_monthly['Sales_shifted'] = data_monthly['Sales'].shift(periods = 1,freq='M')
```
